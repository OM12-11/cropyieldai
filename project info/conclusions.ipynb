# Phase 11: Model Performance Analysis and Improvement

## 1. Objective

The objective of this phase was to evaluate different regression approaches for India-focused crop-yield prediction, compare their performance across six crops, validate their behavior over chronological periods, and select a suitable baseline model for deployment.

## 2. Experiments Conducted

Five model approaches were compared:

1. Shared Linear Regression.
2. Crop-Specific Linear Regression using Year.
3. Crop-Specific Linear Regression using Year and Temperature.
4. Crop-Specific Random Forest.
5. Crop-Specific Polynomial Regression.

A crop-specific mean baseline was also examined. Rolling-origin validation was conducted to compare Year Only and Year + Temperature models across three chronological folds.

## 3. Main Findings

The aggregate experimental comparison favored crop-specific linear models over the shared linear model. Year Only achieved an MAE of approximately 4,074.31 hg/ha, compared with 7,575.83 hg/ha for Shared Linear Regression.

Year + Temperature achieved a slightly lower aggregate MAE of approximately 4,035.69 hg/ha. However, rolling-origin validation did not show a consistent improvement from adding temperature. Year Only achieved lower MAE in the first two folds, while Year + Temperature performed slightly better in the third fold.

Random Forest and Polynomial Regression produced higher aggregate errors than the leading linear approaches in the reported experiments.

## 4. Crop-Wise Findings

During the 2010–2013 validation period, potatoes had the highest absolute MAE at approximately 22,119.10 hg/ha. Soybeans had the highest relative error at approximately 15.46%.

Wheat had the lowest relative error at approximately 4.40%, followed by sweet potatoes at approximately 4.80%.

These findings show that model performance varies by crop and that absolute error alone is insufficient for comparing crops with different yield scales.

## 5. Selected Model

The deployed model remains Crop-Specific Linear Regression using Year as the input feature.

This model was selected because it is simple, interpretable, computationally efficient, and performed competitively across the chronological validation experiments.

The model should be treated as a baseline forecasting approach rather than a comprehensive agricultural prediction system.

## 6. Limitations

* Each crop has only 23 annual observations in the processed dataset, with 2003 missing.
* The models rely on historical yield trends and cannot directly respond to future changes in weather, farming practices, pests, or other agricultural conditions.
* Rainfall is constant at 1083.0 mm across the processed observations and therefore does not provide useful variation for learning rainfall effects.
* Adding temperature did not consistently improve validation performance.
* The reported aggregate model comparison was used during model experimentation and should not be presented as an independent, untouched final-test evaluation.
* Rolling-origin validation covers historical periods only and does not guarantee reliable predictions beyond the observed data range.
* The existing summary metrics do not fully explain the direction or year-specific causes of prediction errors.

## 7. Future Improvements

1. Perform year-by-year residual analysis for potatoes and soybeans.
2. Compare all candidate models against a matched crop-specific baseline using identical chronological folds.
3. Investigate reliable, varying agricultural predictors if suitable data can be obtained.
4. Evaluate future-year predictions against newly observed yields when those data become available.
5. Report MAE, RMSE, relative error, and crop-wise results together rather than relying on a single metric.

## 8. Conclusion

Phase 11 established a baseline model selection supported by aggregate experiments and rolling-origin validation. Crop-specific Year Only Linear Regression remains the deployed model because of its simplicity and competitive performance.

The results also identify potatoes and soybeans as priorities for further error investigation. Predictions for future years should be communicated with appropriate uncertainty because the models extrapolate historical trends and have not been validated against future observations.
